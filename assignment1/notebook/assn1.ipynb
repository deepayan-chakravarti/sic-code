Question 7

```python
from scipy.spatial import distance

a = np.array([5, 4, 3, 2])
b = np.array([1, 2, 5, 5])
c = np.array([4, 4, 4, 3])

euclid_ab = distance.euclidean(a, b)
euclid_bc = distance.euclidean(b, c)
euclid_ac = distance.euclidean(a, c)

print(f"Euclidean Distances - a-b: {euclid_ab:.2f}, b-c: {euclid_bc:.2f}, a-c: {euclid_ac:.2f}")

cos_ab = distance.cosine(a, b)
cos_bc = distance.cosine(b, c)
cos_ac = distance.cosine(a, c)

print(f"Cosine Distances - a-b: {cos_ab:.2f}, b-c: {cos_bc:.2f}, a-c: {cos_ac:.2f}")
```

```python
import sympy as sp

x, k, a, b = sp.symbols('x k a b')

f_A = x**3 - 1
diff_A = sp.diff(f_A, x)
print(diff_A)

f_B = sp.log(x**2 - 3*k)
diff_B = sp.diff(f_B, x)
print(diff_B)

f_C = sp.exp(a * x**b)
diff_C = sp.diff(f_C, x)
print(diff_C)
```

Question 12

```python
x, y, k, a, b = sp.symbols('x y k a b')

f_12 = sp.exp(x**2 + 2*y**2)
print("f_x:", sp.diff(f_12, x))
print("f_y:", sp.diff(f_12, y))
print("f_xx:", sp.diff(f_12, x, x))
print("f_yy:", sp.diff(f_12, y, y))
print("f_xy:", sp.diff(f_12, x, y))
print("f_yx:", sp.diff(f_12, y, x))
```
